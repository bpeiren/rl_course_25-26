# WPO 5: POMDP

In reality, the state of a MDP is often not complete observable. A partially-observed MDP (POMDP) is the result of imperfect measurements (eg. coverage of the sensors) or the state is obscured by design (eg. playing card games). 

In this session, we will experiment on the Battleship POMDP to train an agent to defeat another player. More information the environment can be found here: https://popgym.readthedocs.io/en/latest/autoapi/popgym/envs/battleship/index.html


You can choose an RL agent of your choice (tabular or neural). The challenge is to come up with enhancements that can improve the training of the agent so it can defeat the adversary player. 


## Excercise 1

```python
import popgym

env = popgym.envs.battleship.BattleshipEasy()


## Random agent
terminated, truncated= False, False
obs, _ = env.reset()
n=0
while not (terminated or truncated):
    action = env.action_space.sample()
    next_obs, reward, terminated, truncated, info = env.step(action)
    print(f"Observation in step {n}: {obs}")
    print(f"  Taking action {action}")
    print(f"  Recieving reward {reward}")
    env.render()
    print()
    
    if terminated or truncated:
        break
    obs = next_obs
    n += 1
```
